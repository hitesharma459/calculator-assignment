Ecommerce Sales Analysis (Pandas GroupBy)
1️⃣ Import Libraries and Create Dataset

```python
# Import libraries
import pandas as pd
import random

# For reproducibility (same random data every time)
random.seed(42)

# Define data parameters
regions = ['North', 'South', 'East', 'West']
categories = ['Electronics', 'Clothing', 'Home & Garden', 'Sports', 'Books']
salespersons = ['Alice', 'Bob', 'Carol', 'David', 'Emma', 'Frank']

# Generate synthetic dataset with 200 transactions
data = {
    'transaction_id': range(1001, 1201),
    'region': [random.choice(regions) for _ in range(200)],
    'category': [random.choice(categories) for _ in range(200)],
    'salesperson': [random.choice(salespersons) for _ in range(200)],
    'sales_amount': [round(random.uniform(50, 5000), 2) for _ in range(200)],
    'customer_id': [random.randint(5000, 5100) for _ in range(200)]
}

# Create DataFrame
df = pd.DataFrame(data)

# View dataset
print(df.head(10))

# Shape of dataset
print("\nDataset shape:", df.shape)
```

Task 1: Basic Grouping and Single Aggregations
1️⃣ Total sales for each region

# Group by region and calculate total sales
region_sales = df.groupby('region')['sales_amount'].sum().reset_index()

print(region_sales)

2️⃣ Number of transactions for each category

```python
# Count number of transactions per category
category_transactions = df.groupby('category')['transaction_id'].count().reset_index()

print(category_transactions)
```

3️⃣ Average sales per salesperson


```python
# Calculate average sales for each salesperson
avg_sales_person = df.groupby('salesperson')['sales_amount'].mean().reset_index()

print(avg_sales_person)
```

```python
# Group by region and calculate total sales
region_sales = df.groupby('region')['sales_amount'].sum().reset_index()

print(region_sales)
```

4️⃣ Sort regions by highest sales

```python
# Sort region sales descending
top_region = region_sales.sort_values(by='sales_amount', ascending=False)

print(top_region)
```

Task 2: Multi-Column Grouping and Multiple Aggregations
1️⃣ Total sales by region and category

```python
region_category_sales = df.groupby(['region', 'category'])['sales_amount'].sum().reset_index()

print(region_category_sales)
```

2️⃣ Multiple metrics for each salesperson

```python
salesperson_stats = df.groupby('salesperson')['sales_amount'].agg(['sum', 'mean', 'count']).reset_index()

print(salesperson_stats)
```

3️⃣ Identify top salesperson

```python
top_salesperson = salesperson_stats.sort_values(by='sum', ascending=False)

print(top_salesperson)
```

4️⃣ Category with highest revenue

```python
# Total revenue by category
category_sales = df.groupby('category')['sales_amount'].sum()

# Category with max revenue
top_category = category_sales.idxmax()

print("Category with highest revenue:", top_category)
```

Task 3: Custom Aggregation and Complete Sales Report
1️⃣ Custom function (Sales Range)

```python
# Custom aggregation function
def sales_range(x):
    return x.max() - x.min()
```

2️⃣ Apply custom + standard aggregations

```python
region_analysis = df.groupby('region')['sales_amount'].agg(
    ['sum', 'mean', 'min', 'max', sales_range]
).reset_index()

print(region_analysis)
```

3️⃣ Final Sales Summary Report

```python
final_report = df.groupby('region').agg({
    'sales_amount': ['sum', 'mean'],   # total and average sales
    'customer_id': 'count'             # number of transactions
}).reset_index()

print(final_report)
```
